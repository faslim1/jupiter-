```python
amazon_df['rating_count'] = pd.to_numeric(amazon_df['rating_count'], errors='coerce')
amazon_df['rating_count'] = amazon_df['rating_count'].fillna(amazon_df['rating_count'].median())
amazon_df['rating_count'].dtype
```

```python
amazon_df['discounted_price'] = amazon_df['discounted_price'].str.replace(r'[₹ ,]','', regex = True)
amazon_df['discounted_price'] = pd.to_numeric(amazon_df['discounted_price'], errors = 'coerce')
```

```python
amazon_df['discount_percentage'] = amazon_df['discount_percentage'].str.replace('%','',regex = False)
amazon_df['discount_percentage'] = pd.to_numeric(amazon_df['discount_percentage'], errors= 'coerce')
amazon_df['discount_percentage'].dtype
```

```python
amazon_df['category'] = amazon_df['category'].str.strip().str.replace(r'\s+' ,' ' , regex = True)
amazon_df[['category']]
```

```python
amazon_df['category'] = amazon_df['category'].str.replace('&',' & ', regex= False)
amazon_df['category']
```

```python
category_split = amazon_df['category'].str.split('|' , expand = True)
category_split
```

```python
amazon_df['main_category'] = category_split[0]

amazon_df['sub_category'] = category_split[1]

amazon_df['sub_sub_category'] = category_split[2]
```
