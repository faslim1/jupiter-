```python
import pandas as pd
```

1. Load the dataset.

```python
df = pd.read_csv('googleplaystore.csv')
```

2. Display the first 5 rows.

```python
df.head()
```

3. Display the last 5 rows

```python
df.tail()
```

4. Check dataset structure and non-null values

```python
df.info()
```

5. Summary statistics for all columns

```python
df.describe(include="all")
```

6. Display column names and data types

```python
print(df.columns)
print(df.dtypes)
```

7. Identify missing values in each column

```python
df.isnull().sum()
```

8. Remove rows where Rating is missing

```python
df = df.dropna(subset=["Rating"])
```

9. Fill missing values in Type with the most frequent value

```python
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
```

10. Remove duplicate records

```python
df = df.drop_duplicates()
```

11. Clean Installs and convert to numeric

```python
df["Installs"] = df["Installs"].str.replace("+", "", regex=False)
df["Installs"] = df["Installs"].str.replace(",", "", regex=False)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")
```

12. Clean Price and convert to numeric

```python
df["Price"] = df["Price"].str.replace("$", "", regex=False)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
```

13. Convert Reviews to integer

```python
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

df = df.dropna(subset=["Reviews"])

df["Reviews"] = df["Reviews"].astype(int)
```

14. Convert Last Updated to datetime

```python
df["Last Updated"] = pd.to_datetime(
    df["Last Updated"],
    errors="coerce"
)
```

15. Convert all app names to lowercase

```python
df["App"] = df["App"].str.lower()
```

16. Convert Category names to uppercase

```python
df["Category"] = df["Category"].str.upper()
```

17. Split Genres into multiple parts

```python
df["Genres"].str.split(";")
```

18. Replace unwanted text patterns in Category

```python
df["Category"] = df["Category"].str.replace("_", " ", regex=False)
```

19. Filter apps with Rating > 4.5

```python
high_rating_apps = df[df["Rating"] > 4.5]

high_rating_apps
```

20. Find all free apps

```python
free_apps = df[df["Price"] == 0]

free_apps
```

21. Apps with more than 1,000,000 installs

```python
popular_apps = df[df["Installs"] > 1000000]

popular_apps
```

22. Select apps belonging to multiple categories

```python
selected_apps = df[
    df["Category"].isin(
        ["GAME", "FAMILY", "TOOLS"]
    )
]

selected_apps
```

23. Filtering using query()

```python
df.query("Rating > 4.5")
```

24. Create Revenue column

```python
df["Revenue"] = df["Installs"] * df["Price"]

df.head()
```

25. Create Rating Groups: Low, Medium, High

```python
df["Rating_Group"] = pd.cut(
    df["Rating"],
    bins=[0, 3, 4, 5],
    labels=["Low", "Medium", "High"]
)

df.head()
```

26. Flag high-install apps (>1 million)

```python
df["High_Install"] = df["Installs"].apply(
    lambda x: 1 if x > 1000000 else 0
)

df.head()
```

27. Convert Type into numeric values

```python
df["Type_Numeric"] = df["Type"].map(
    {
        "Free": 0,
        "Paid": 1
    }
)

df.head()
```

```python
df.head()
```

28. Sort apps by installs in descending order

```python
df.sort_values(
    by="Installs",
    ascending=False
)
```

29. Top 10 apps based on Reviews

```python
df.nlargest(10, "Reviews")
```

30. Bottom 10 apps based on installs

31. Rank apps based on Rating

```python
df["Rating_Rank"] = df["Rating"].rank(
    ascending=False
)

df[["App", "Rating", "Rating_Rank"]]
```

32. Average Rating for each Category

```python
df.groupby("Category")["Rating"].mean()
```

33. Total installs for each Category

```python
df.groupby("Category")["Installs"].sum()
```

34. Count number of apps in each Category

```python
df["Category"].value_counts()
```

35. Multiple aggregations per Category

```python
df.groupby("Category").agg(
    {
        "Rating": "mean",
        "Installs": "max"
    }
)
```

36. Split dataset into multiple tables

```python
apps_info = df[
    ["App", "Category", "Rating"]
]

installs_info = df[
    ["App", "Installs", "Reviews"]
]

pricing_info = df[
    ["App", "Type", "Price"]
]
```

```python
apps_info.head()
```

```python
installs_info.head()
```

```python
pricing_info.head()
```

37. Merge app details with installs data

```python
merged_df = pd.merge(
    apps_info,
    installs_info,
    on="App"
)

merged_df.head()
```

38. Perform an INNER JOIN

```python
inner_join = pd.merge(
    apps_info,
    installs_info,
    on="App",
    how="inner"
)

inner_join.head()
```

39. Perform LEFT JOIN and find missing values

```python
left_join = pd.merge(
    apps_info,
    pricing_info,
    on="App",
    how="left"
)

left_join.head()
```
