```python
import matplotlib.pyplot as plt
plt.subplot(2,2,1)
plt.plot([1,2,3],[4,5,6],marker='o')
plt.title('Plot 1')

plt.subplot(2,2,2)
plt.plot([1,2,3],[6,5,4],marker='s')
plt.title('plot 2')

plt.subplot(2,2,3)
plt.plot([2,4,6],[1,2,5],marker='^')
plt.title('plot 3')

plt.subplot(2,2,4)
plt.plot([3,2,5],[6,1,4],marker='v')
plt.title('plot 4')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.subplot(2,2,1)
plt.plot([0,1,2,3],[10,1,8,3],marker='o')
plt.title('Plot 1')

plt.subplot(2,2,2)
plt.plot([3,2,1,0],[10,20,30,40],marker='s')
plt.title('plot 2')

plt.subplot(2,2,3)
plt.plot([0,1,2,3],[40,30,20,10],marker='^')
plt.grid()

plt.subplot(2,2,4)
plt.plot([3,2,5],[10,1,8,3],marker='v')
plt.title('plot 4')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
x = np.arange(5)
print(x)

fig,axis = plt.subplots(1,2)

axis[0].plot(x,x**2,marker='o',markerfacecolour='red')
axis[0].set_title('Square')
axis[0].set_xlabel('Numbers')
axis[0].set_ylabel('Square of numbers')

axis[1].plot(x,x**3,marker='s',markerfacecolour='red')
axis[1].set_title('cube')
axis[1].set_xlabel('Numbers')
axis[1].set_ylabel('Cube of numbers')

plt.tight_layout()
plt.show()

```
