31. Convert the Sales column to a NumPy array and find: maximum, minimum, mean, median,
and standard deviation.

```python
sales_array = df["Sales"].to_numpy()

print("Maximum:", np.max(sales_array))
print("Minimum:", np.min(sales_array))
print("Mean:", np.mean(sales_array))
print("Median:", np.median(sales_array))
print("Standard deviation:", np.std(sales_array))
```

33. Using NumPy, create a normalized Sales column using (Sales - mean)/std.

```python
sales_array = df["Sales"].to_numpy()

df["Normalized Sales"] = (
    sales_array - np.mean(sales_array)
) / np.std(sales_array)
```
